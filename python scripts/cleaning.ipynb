```python
import pandas as pd

events = pd.read_excel("care_flow_event_log.xlsx")
cases = pd.read_excel("care_flow_case_details.xlsx")

print(events.head(10))
print(cases[["Case_ID", "Arrival_Timestamp"]].head())
```

```python
events.groupby("Case_ID")["Event_Order"].apply(
    lambda s: s.is_monotonic_increasing
).value_counts()
```

```python
# Convert time-only values to timedeltas
events["Event_Time"] = pd.to_timedelta(events["Timestamp"].astype(str))
cases["Arrival_Time"] = pd.to_timedelta(cases["Arrival_Timestamp"].astype(str))

# Add arrival time to each event
events = events.merge(
    cases[["Case_ID", "Arrival_Time"]],
    on="Case_ID",
    how="left",
    validate="many_to_one"
)

# Sort by patient and event sequence
events = events.sort_values(
    ["Case_ID", "Event_Order"]
).reset_index(drop=True)

# Detect midnight crossings within each case
previous_time = events.groupby("Case_ID")["Event_Time"].shift()
crossed_midnight = events["Event_Time"] < previous_time

events["Day_Offset"] = (
    crossed_midnight.groupby(events["Case_ID"])
    .cumsum()
    .fillna(0)
    .astype(int)
)

# Build reconstructed timestamps relative to the arrival date
events["Timestamp_Reconstructed"] = (
    pd.Timestamp("2026-01-01")
    + events["Day_Offset"] * pd.Timedelta(days=1)
    + events["Event_Time"]
    - events["Arrival_Time"]
)

# Inspect the first case
display(
    events.loc[events["Case_ID"] == "CF00001", [
        "Case_ID", "Activity_Name", "Event_Order",
        "Timestamp", "Day_Offset", "Timestamp_Reconstructed"
    ]]
)
```

```python
#Calculate patient journey duration
journey = (
    events.groupby("Case_ID")["Timestamp_Reconstructed"]
    .agg(["min", "max"])
    .reset_index()
)

journey["Journey_Duration"] = (
    journey["max"] - journey["min"]
)

journey.head(10)
```

```python

journey["Journey_Duration"].describe()
```

```python

import pandas as pd

# ==========================================
# 1. COUNT ACTIVITIES PER PATIENT
# ==========================================

# Group by patient and activity to count occurrences
activity_counts = events.groupby(
    ["Case_ID", "Activity_Name"]
).size().reset_index(name="Count")

# Identify activities performed more than once
repeated = activity_counts[activity_counts["Count"] > 1]


# ==========================================
# 2. CALCULATE REPEAT ACTIVITY RATE
# ==========================================

# Count unique patient cases
total_cases = events["Case_ID"].nunique()

# Count patients who repeated at least one activity
repeat_cases = repeated["Case_ID"].nunique()

# Calculate percentage of patients with repeated activities
repeat_rate = repeat_cases / total_cases * 100


# ==========================================
# 3. CALCULATE PATIENT JOURNEY DURATION
# ==========================================

# Find the first and last reconstructed timestamp
# for each patient
journey = events.groupby("Case_ID")[
    "Timestamp_Reconstructed"
].agg(Start="min", End="max")

# Calculate total journey duration in minutes
journey["Duration_Min"] = (
    journey["End"] - journey["Start"]
).dt.total_seconds() / 60


# ==========================================
# 4. CALCULATE TIME BETWEEN ACTIVITIES
# ==========================================

# Arrange events in patient activity order
events = events.sort_values(
    ["Case_ID", "Event_Order"]
).copy()

# Get the previous activity for each patient
events["Previous_Activity"] = events.groupby(
    "Case_ID"
)["Activity_Name"].shift()

# Get the previous event's timestamp
events["Previous_Time"] = events.groupby(
    "Case_ID"
)["Timestamp_Reconstructed"].shift()

# Calculate minutes between consecutive activities
events["Transition_Min"] = (
    events["Timestamp_Reconstructed"]
    - events["Previous_Time"]
).dt.total_seconds() / 60

# Calculate the count, average and median time
# for each activity-to-activity transition
transitions = (
    events.dropna(subset=["Previous_Activity"])
    .groupby(["Previous_Activity", "Activity_Name"])
    ["Transition_Min"]
    .agg(["count", "mean", "median"])
    .reset_index()
)


# ==========================================
# 5. DISPLAY KPI RESULTS
# ==========================================

# Total number of unique patients
print("Total Patients:", total_cases)

# Total number of recorded hospital activities
print("Total Events:", len(events))

# Number of patients with repeated activities
print("Patients with Repeated Activities:", repeat_cases)

# Percentage of patients with repeated activities
print(f"Repeat Activity Rate: {repeat_rate:.2f}%")

# Average time spent in the hospital journey
print(
    f"Average Journey Duration: "
    f"{journey['Duration_Min'].mean():.2f} minutes"
)

# Median time spent in the hospital journey
print(
    f"Median Journey Duration: "
    f"{journey['Duration_Min'].median():.2f} minutes"
)

# Average time gap between consecutive events
print(
    f"Average Transition Gap: "
    f"{events['Transition_Min'].mean():.2f} minutes"
)


# ==========================================
# 6. DISPLAY ANALYSIS TABLES
# ==========================================

# Show the first five patient journey records
display(journey.head())

# Show the first ten repeated activity records
display(repeated.head(10))

# Show the ten transitions with the highest
# average time gap
display(
    transitions.sort_values(
        "mean", ascending=False
    ).head(10)
)
```

```python

import os

# Use the existing output folder
os.makedirs("output of pandas", exist_ok=True)

# Save each analysis result as an Excel file
events.to_excel(
    "output of pandas/careflow_cleaned_events.xlsx",
    index=False
)

journey.to_excel(
    "output of pandas/careflow_journey_summary.xlsx",
    index=True
)

repeated.to_excel(
    "output of pandas/careflow_repeated_activities.xlsx",
    index=False
)

transitions.to_excel(
    "output of pandas/careflow_transition_analysis.xlsx",
    index=False
)

print("All 4 Excel files exported successfully!")
```

```python

# Merge patient details with journey duration
patient_summary = cases.merge(
    journey[["Duration_Min"]],
    left_on="Case_ID",
    right_index=True,
    how="left",
    validate="one_to_one"
)

# Check the merged data
print("Rows:", patient_summary.shape[0])
print("Columns:", patient_summary.shape[1])

display(patient_summary.head())
```

```python
# Count distinct activity types repeated for each patient
repeat_summary = (
    repeated.groupby("Case_ID")
    .agg(
        Repeated_Activity_Types=("Activity_Name", "nunique"),
        Total_Repeated_Events=("Count", "sum")
    )
    .reset_index()
)

# Add repeat information to the patient summary
patient_summary = patient_summary.merge(
    repeat_summary,
    on="Case_ID",
    how="left",
    validate="one_to_one"
)

# Patients without repeated activities get zero
patient_summary[
    ["Repeated_Activity_Types", "Total_Repeated_Events"]
] = patient_summary[
    ["Repeated_Activity_Types", "Total_Repeated_Events"]
].fillna(0).astype(int)

display(patient_summary.head())
```

```python
# Save the final patient-level dataset
patient_summary.to_excel(
    "output of pandas/careflow_patient_summary.xlsx",
    index=False
)

print("Patient summary exported successfully!")
```
